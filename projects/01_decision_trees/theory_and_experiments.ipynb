****Overview of Decision Trees:****

**What are decison trees?**

- Supervised Learning algorithm that can be used for classification or regression
- Makes predictions by recursively splitting data into smaller groups based on predictor values

**Components of a decision tree:**

- Root Node - First node, containing all training observations
- Internal Nodes - Represent attribute tests
- Branches - represent outcomes of attribute tests (True or False)
- Leaf Nodes - represent final decisions or predictions

****Mathematics Behind the Model:****

**Impurity measures (finding the best split):**

**Gini**

$$Gini : 1 - \sum_{i=1}^{n} p_i^2$$

- Measures how mixed the observations/classes are within a node
- lower gini indicates greater purity
- Gini = 0 indicates a perfectly pure node - we want as pure as a node as possible

**Entropy**

$$Entropy : H = - \sum_{k=1}^{K} p_k \log_2(p_k)$$

- Measures how mixed classes are witin a node
- high entropy - classes are evenly distributed (high impurity)
- entropy = 0 - all observations belong to the same class (perfect purity)

*Goal for our model - Have a gini index, and entropy as close to 0 as possible.*

****How do decision trees relate to Random Forests?****

Random forests are a combination of many deicison trees used to make a final prediction.
- train many trees 
- each tree learns from a random sample of data
- each tree selects a random subset of features 
- each tree makes a prediction
- combine the predictions and pick the majority prediction out of the trees.
